This notebook validates the deterministic sequence-analysis engine using an explicit offline test fixture. The fixture is synthetic test data and is **not** presented as the NCBI biological reference sequence.

The production path remains `scripts/reference_sequence_analysis.py`, which retrieves the versioned NCBI RefSeq record `NM_004972.4` at runtime.

## Canonical variant

`NM_004972.4:c.1849G>T → p.Val617Phe (V617F)`; dbSNP `rs77375493`; GRCh38 `NC_000009.12:g.5073770G>T`.

```python
from pathlib import Path
import sys
import pandas as pd
ROOT = Path.cwd().parent if Path.cwd().name == 'analysis' else Path.cwd()
sys.path.insert(0, str(ROOT / 'scripts'))
from reference_sequence_analysis import analyze_codon

# Deterministic synthetic fixture: GTC at c.1849-c.1851.
sequence = 'A' * 1848 + 'GTC' + 'A' * (7023 - 1851)
print(f'Offline validation fixture length: {len(sequence):,} nt')
```

```python
result = analyze_codon(sequence)
pd.Series(result)
```

```python
assert result['hgvs_c'] == 'NM_004972.4:c.1849G>T'
assert result['reference_codon'] == 'GTC'
assert result['alternate_codon'] == 'TTC'
assert result['reference_amino_acid'] == 'V'
assert result['alternate_amino_acid'] == 'F'
assert result['hgvs_p'] == 'p.Val617Phe'
print('PASS: NM_004972.4:c.1849G>T')
print('PASS: GTC (Val) → TTC (Phe)')
print('PASS: p.Val617Phe')
```
