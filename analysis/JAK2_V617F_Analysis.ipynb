# JAK2 V617F Molecular Analysis

Reproducible educational analysis of **JAK2 NM_004972.4:c.1849G>T (p.Val617Phe)**. ClinVar records place the variant at GRCh38 chr9:5073770 and link it to rs77375493. citeturn0search0turn0search1


## Objectives

- Verify the canonical variant representation.
- Create a machine-readable annotation table.
- Retrieve public NCBI variation evidence.
- Summarize condition-specific evidence without conflating somatic and germline assertions.
- Visualize the molecular interpretation pathway.

> Educational/research workflow; not a clinical diagnostic assay.


```python
from dataclasses import dataclass
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
```

```python
@dataclass(frozen=True)
class VariantRecord:
    gene: str = 'JAK2'
    transcript: str = 'NM_004972.4'
    hgvs_c: str = 'NM_004972.4:c.1849G>T'
    hgvs_p: str = 'p.Val617Phe'
    dbsnp: str = 'rs77375493'
    assembly: str = 'GRCh38'
    genomic_hgvs: str = 'NC_000009.12:g.5073770G>T'
    domain: str = 'JH2 pseudokinase domain'

variant = VariantRecord()
assert variant.hgvs_p == 'p.Val617Phe'
variant
```

```python
annotation = pd.DataFrame([vars(variant)])
annotation.T.rename(columns={0:'value'})
```

## Nucleotide → protein consequence

The defining substitution is **G>T at cDNA position 1849**, producing **Val617Phe (V617F)**. ClinVar describes the consequence as a missense variant. citeturn0search0


```python
consequence = pd.DataFrame({
    'Level': ['Nucleotide', 'Protein', 'Functional domain'],
    'Reference': ['G', 'Val (V)', 'JH2 pseudokinase domain'],
    'Alternate': ['T', 'Phe (F)', 'Altered regulatory function'],
})
consequence
```

## Public evidence summary

ClinVar contains condition-specific records for polycythemia vera, primary myelofibrosis and thrombocythemia. The records also illustrate why variant interpretation must preserve **disease context and variant origin** rather than treating every database classification as interchangeable. citeturn0search0turn0search1turn0search2


```python
evidence = pd.DataFrame([
    {'Condition':'Acquired polycythemia vera','ClinVar_context':'Pathogenic; multiple submissions','Origin_note':'Somatic JAK2 V617F is a recognized MPN driver'},
    {'Condition':'Primary myelofibrosis','ClinVar_context':'Pathogenic; multiple submissions','Origin_note':'Somatic JAK2 V617F occurs in a substantial subset'},
    {'Condition':'Thrombocythemia 3','ClinVar_context':'Likely pathogenic; 2 submissions','Origin_note':'Interpret within the appropriate disease context'},
])
evidence
```

```python
fig, ax = plt.subplots(figsize=(10, 2.8))
steps = ['DNA variant','c.1849G>T','p.Val617Phe','JH2 domain','JAK-STAT dysregulation','MPN context']
ax.plot(range(len(steps)), [1]*len(steps), marker='o')
for i, step in enumerate(steps):
    ax.text(i, 1.02, step, ha='center', va='bottom', rotation=20)
ax.set_yticks([])
ax.set_xticks([])
ax.set_title('JAK2 V617F molecular interpretation pathway')
ax.spines[['left','right','top']].set_visible(False)
plt.tight_layout()
plt.show()
```
